```python
# Practice Questions

class car:
    def __init__(self,brand,model,speed=0):
        self.brand=brand
        self.model=model
        self.speed=speed

    def accerlate(self):
        self.speed+=10
        print(f"speed after accerlate {self.speed }")
    
    def brake(self):
        if self.speed - 10 < 0:
            self.speed=0

        self.speed-=10
        print(f"speed after brake {self.speed}")

    def display(self):
        print(f"Brand name {self.brand }, model {self.model}, speed is {self.speed}")

C=car("Toyota","2025")
C.accerlate()
C.accerlate()
C.brake()
C.brake()
C.display()
```

```python
# Inhertiance and polymorfism
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    
    def introduce(self):
        print(f"MY name is {self.name },Age is {self.age}")
    
class Student(Person):
    def __init__(self,name,age,student_id):
        super().__init__(name,age)
        self.student_id=student_id

    def introduce(self):
        print(f"name is {self.name}, age is {self.age}, student id is {self.student_id}")


p=Person("Mustafa",20)
s=Student("Ali",22,180)

for x in (p,s):
    x.introduce()
```

```python
class BankAccount:
    def __init__(self):
        self.__balance=0

    def deposit(self,amount):
        if amount >0:
            self.__balance+=amount
        else:
            print("Sorry")

    def withdraw(self,amount):
        if self.__balance>amount:
            self.__balance-=amount
    
    def get_balance(self):
        return self.__balance
    
c=BankAccount
```
